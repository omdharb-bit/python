#### Pandas-DataFrame And Series
Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
# Series
##A Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
# Create a series from dictioary elemnts
data= {'a':1,'b':2,'c':3}
series_dictionary=pd.Series(data)
print(series_dictionary)
```

```python
data=[10,20,30,]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
# dataFrame
# create a datarame from a dictionary of unit

data={
'Name':['Om','Ram','Dhruv'],
'Age':[19,20,19],
'City':['Jabalpur','Indore','Bhopal']

}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
# Create a dataframe from list of dictionary

data=[
 { 'Name':'Om','Age':19,'City':'Jabalpur'},
 { 'Name':'Dhruv','Age':18,'City':'Bhopal'},
 { 'Name':'Hittu','Age':19,'City':'Indore'},
 { 'Name':'Ram','Age':17,'City':'Sihora'}
]
df = pd.DataFrame(data)
print(df)
print(type(df))
```
