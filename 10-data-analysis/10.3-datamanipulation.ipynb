Data Manipulation and analysis with pandas

Definition of Data Manipulation & Analysis in Pandas

Data Manipulation and Analysis with Pandas means:

Using the Pandas library in Python to clean, organize, transform, and analyze data in order to extract useful information and insights.

🔍 In Simple Words

👉 It is the process of:

Manipulation = modifying data (cleaning, filtering, sorting, adding columns)
Analysis = understanding data (finding patterns, averages, trends)

```python
# Rename columms
df = df.rename(columns={'Date': 'Order Date'})
df.head()
```

```python
# Change Data types

df['skills_count_new']=df['skills_count'].fillna(df['skills_count'].mean()).astype(int)
df.head()
```

```python
df["skills_count_new"] = df["skills_count"].apply(lambda x: x * 2)
df.head()
```

```python
# Data Aggregating and grouping

df.head()
```

```python
grouped_mean=df.groupby('industry')['skills_count'].mean()
print(grouped_mean)
```
