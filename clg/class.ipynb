```python
lst= [[1,3,5,6],[3,4,5],[3,5,7],[10,20,30]]
# output=[15,12,15,60]

lst_sum=[]

for i in lst:
  total=0
  for j in i:
    total+=j
  lst_sum.append(total)
print(lst_sum)
```

```python
lst = [[1, 3, 5, 6], [3, 4, 5], [3, 5, 7], [10, 20, 30]]
ans = []

for i in range(len(lst)):  # now i is index
    result = lst[i].copy()  # avoid modifying original list

    if i % 2 == 0:  # even index
        for j in range(len(result)):
            if result[j] % 2 == 0:
                result[j] -= 1
    else:  # odd index
        for j in range(len(result)):
            if result[j] % 2 != 0:
                result[j] += 1

    ans.append(result)

print(ans)
```
