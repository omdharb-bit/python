```python
lst = [[1, 3, 5, 6], [3, 4, 5], [3, 5, 7], [10, 20, 30]]
ans = []

for i in range(len(lst)):  # now i is index
    result = lst[i].copy()  # avoid modifying original list

    if i % 2 == 0:  # even index
        for j in range(len(result)):
            if result[j] % 2 == 0:
                result[j] -= 1
    else:  # odd index
        for j in range(len(result)):
            if result[j] % 2 != 0:
                result[j] += 1

    ans.append(result)
print(ans)
```

```python
# Write a programme to calculate and print sum of max elements of each list of lists
lst=[[1,2,4,5],[3,5,4,3],[4,5,3,2]]
max_elements = [max(sublist) for sublist in lst]
print(sum(max_elements))
```

```python
# wap to claculate and point the summ of elemments in each list of lists and add the resultant values.Without using sum function

lst=[[1,2,4,5],[3,5,4,3],[4,5,3,2]]
total = 0
for sublist in lst:
    sublist_sum = 0
    for num in sublist:
        sublist_sum += num
    total += sublist_sum
print(total)
```
